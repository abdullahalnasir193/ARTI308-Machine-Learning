```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

#read datasets
df = pd.read_csv('global_cars_enhanced.csv')
```

Task 1: Identify data quality issues

```python
print(df.isna().sum())

print(df.info())
```

Task 2: Apply one missing value strategy

```python
# 1. Introduce artificial missing values (since the original data is clean)
# Selecting the first 6 rows of 'Price_USD' to make them NaN
df.loc[0:5, 'Price_USD'] = np.nan

# 2. Apply Mean Imputation strategy
# Reason: Price_USD is a numerical variable, and mean imputation preserves the average
imputer = SimpleImputer(strategy='mean')
df['Price_USD'] = imputer.fit_transform(df[['Price_USD']])
```

Task 3 - Detect and Handle Outliers (IQR)

```python
# Focus on 'Horsepower' column for outlier detection
Q1 = df['Horsepower'].quantile(0.25)
Q3 = df['Horsepower'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers from the dataset
df_clean = df[(df['Horsepower'] >= lower_bound) & (df['Horsepower'] <= upper_bound)]

```

Task 4 - Normalize Numerical Features

```python
# Select columns to scale
cols_to_scale = ['Engine_CC', 'Horsepower', 'Price_USD']

# Method 1: Min-Max Scaling (Scales data between 0 and 1)
min_max_scaler = MinMaxScaler()
df_minmax = df_clean.copy()
df_minmax[cols_to_scale] = min_max_scaler.fit_transform(df_clean[cols_to_scale])

# Method 2: Z-score Normalization (Standardization: Mean=0, Std=1)
std_scaler = StandardScaler()
df_zscore = df_clean.copy()
df_zscore[cols_to_scale] = std_scaler.fit_transform(df_clean[cols_to_scale])

```

Task 5 - Apply PCA and Interpretation

```python
# Apply PCA on the Z-score standardized features
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_zscore[cols_to_scale])

# Get the Explained Variance Ratio
variance_ratio = pca.explained_variance_ratio_

print("Task 5: PCA Results")
print(f"Explained Variance Ratio: {variance_ratio}")
print(f"Total variance captured by 2 components: {sum(variance_ratio)*100:.2f}%")
```
